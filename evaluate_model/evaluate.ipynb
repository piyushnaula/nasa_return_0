```python
import os
```

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\naula\OneDrive\Desktop\Nasa Return 0\dataset\raw\raw_data.csv", comment="#", engine="python")
df.head()
```

```python
drop_cols = [c for c in df.columns if "_err1" in c or "_err2" in c]
df = df.drop(columns=drop_cols, errors="ignore")
```

```python
df.head()
```

```python
drop_id_cols = ["kepid", "kepoi_name", "kepler_name"]
df = df.drop(columns=[c for c in drop_id_cols if c in df.columns])
```

```python
df.head()
```

```python
cat_cols = df.select_dtypes(include=["object"]).columns
for col in cat_cols:
    df[col] = df[col].astype("category").cat.codes
```

```python
df.head()
```

```python
num_cols = df.select_dtypes(include=["float64", "int64"]).columns
df[num_cols] = df[num_cols].apply(lambda x: x.fillna(x.median()))
```

```python
df.head()
```

```python
os.makedirs(os.path.dirname(r"C:\Users\naula\OneDrive\Desktop\Nasa Return 0\dataset\cleaned\cleaned_data.csv"), exist_ok=True)
df.to_csv(r"C:\Users\naula\OneDrive\Desktop\Nasa Return 0\dataset\cleaned\cleaned_data.csv", index=False)
```
